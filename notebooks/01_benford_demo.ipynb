# Digit distributions
An anomaly is not proof of wrongdoing. Bounded amounts and rounding can change digit frequencies. MAD describes the size of the deviation; chi-square tests a sampling model.

```python
import numpy as np
import pandas as pd
from fraudscope.benford import benford_test
from fraudscope.plots import plot_benford
rng = np.random.default_rng(11)
amounts = pd.Series(10 ** rng.uniform(1, 6, 4000), name='amount')
result = benford_test(amounts, 'first')
print(result.to_dict()['details']['conformity'], result.caveats)
plot_benford(result)
```
