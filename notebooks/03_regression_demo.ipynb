# Regression diagnostics
OLS and robust Huber coefficients help identify influential rows; model assumptions matter. An anomaly is not proof of wrongdoing. Outliers and group differences can have ordinary explanations.

```python
import numpy as np
import pandas as pd
from fraudscope.regression import regression_test
from fraudscope.plots import plot_residuals, plot_group_residuals
rng = np.random.default_rng(13)
x = rng.normal(size=200)
df = pd.DataFrame({'x': x, 'y': 2 + 3*x + rng.normal(size=200), 'group': np.tile(['a', 'b'], 100)})
df.loc[8, 'y'] += 10
result = regression_test(df, 'y', ['x'], group='group')
print(result.flagged_indices, result.caveats)
plot_residuals(result)
plot_group_residuals(result)
```
