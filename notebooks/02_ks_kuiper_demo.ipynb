# ECDF tests
KS measures the largest CDF gap. Kuiper sums positive and negative gaps, making it useful for cyclic measurements. An anomaly is not proof of wrongdoing: different collection periods or rounding can create deviations.

```python
import numpy as np
from scipy import stats
from fraudscope.ks import ks_one_sample, ks_two_sample
from fraudscope.kuiper import kuiper_one_sample, kuiper_two_sample
from fraudscope.plots import plot_ecdf_vs_reference
rng = np.random.default_rng(12)
a, b = rng.normal(size=200), rng.normal(loc=0.3, size=200)
ks = ks_one_sample(a, stats.norm())
kuiper = kuiper_one_sample(a, stats.norm())
print(ks.p_value, kuiper.p_value, ks_two_sample(a, b).statistic, kuiper_two_sample(a, b).statistic)
plot_ecdf_vs_reference(ks)
```
